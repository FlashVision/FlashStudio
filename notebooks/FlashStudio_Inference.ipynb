```
# Troubleshooting: If the URL above doesn't work, run this cell instead:
import subprocess, sys, time

proc = subprocess.Popen(
    [sys.executable, "-m", "streamlit", "run",
     __import__('flashstudio.launcher', fromlist=['_get_app_path'])._get_app_path(),
     "--server.port", "8501", "--server.headless", "true",
     "--server.enableCORS", "false", "--server.enableXsrfProtection", "false"],
    stdout=subprocess.PIPE, stderr=subprocess.PIPE
)
time.sleep(5)

from google.colab.output import eval_js
from IPython.display import display, HTML
url = eval_js("google.colab.kernel.proxyPort(8501)")
display(HTML(f'<a href="{url}" target="_blank" style="font-size:16px;padding:10px 24px;background:#7C3AED;color:white;border-radius:8px;text-decoration:none;font-weight:600;">Open FlashStudio</a>'))
```
