# ⚡ FlashStudio — Inference

Run FlashDet object detection on your images with an interactive UI.

**Requirements:** GPU runtime recommended (CPU also works)
**Mobile:** Works on phones & tablets — a public URL is created automatically.

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/FlashVision/FlashStudio/blob/main/notebooks/FlashStudio_Inference.ipynb)

```
# Step 2: Launch FlashStudio
# Creates a public URL that works on mobile & any device — no token needed!
from flashstudio import launch
launch()
```
